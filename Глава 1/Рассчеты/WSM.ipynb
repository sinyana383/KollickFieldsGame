Взвешанные суммы Таблицы 1. Анализ боевых систем

```python
def weight_sum(dic: dict, crits: dict):
    summ = 0
    for k,v in dic.items():
        f = v*crits[k]
        print(f"{k}: {v}, {f}")
        summ += f
    print(f"summ: {summ}")
    print("--------------")
    return summ
    
    
    

crits = {'reviews_n' : 0.2, 'pos_rev_n' : 0.2, 'is_adventure' : 0.25, 'is_unity' : 0.35}
rev_max = 53796
bns = {'reviews_n' : 53796 / rev_max, 'pos_rev_n' : 0.95, 'is_adventure' : 1, 'is_unity' : 1}
skyrim = {'reviews_n' : 8341 / rev_max, 'pos_rev_n' : 0.73, 'is_adventure' : 1, 'is_unity' : 0}
gorn = {'reviews_n' : 7380 / rev_max, 'pos_rev_n' : 0.91, 'is_adventure' : 0, 'is_unity' : 1}


res = {'bns': weight_sum(bns, crits), 'skyrim' : weight_sum(skyrim, crits), 'gorn': weight_sum(gorn, crits)}

for k,v in res.items():
    print(f"{k}: {v}")
```
